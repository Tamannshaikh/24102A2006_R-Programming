## Lecture 18: Missing Data Handling

### 1. Assign a missing value (NA) to a variable

```r
x = NA
x
is.na(x)
```

```r
x = c(11, NA, 13, NA)
x
is.na(x)
```

```r
x = c(11, NA, 13, NA)

mean(x)
mean(x, na.rm = TRUE)
```

## Lecture 20: Nested if-else and ifelse

### 20. Use nested if-else with x greater than 3

```r
x = 5

if (x == 3) {
  x = x - 1
} else if (x < 3) {
  x = x + 5
} else {
  x = 2 * x
}

x
```

### 21. Use nested if-else with x less than 3

```r
x = 2

if (x == 3) {
  x = x - 1
} else if (x < 3) {
  x = x + 5
} else {
  x = 2 * x
}

x
```

### 22. Use nested if-else when x is equal to 3

```r
x = 3

if (x == 3) {
  x = x - 1
} else if (x < 3) {
  x = x + 5
} else {
  x = 2 * x
}

x
```

### 23. Apply ifelse() to a vector of values

```r
x = 1:10
x
ifelse(x < 6, x^2, x + 1)
```

### 24. Identify even and odd numbers using ifelse() and modulo operator

```r
x = c(7, 9, 8, 4)
ifelse(x %% 2 == 0, "even number", "odd number")
```

## Lecture 21: switch() and which() for Conditional Execution

### 25. Use switch() with an integer expression

```r
switch(2, "apple", "banana", "orange")
switch(1, "apple", "banana", "orange")
```

### 26. Use switch() with a character string and named cases

```r
switch("colour", "colour" = "blue", "gender" = "male", "volume" = 50)
switch("volume", "colour" = "blue", "gender" = "male", "volume" = 50)
```

### 27. Use switch() when there is no matching case

```r
switch(4, "apple", "banana", "orange")
switch("size", "colour" = "blue", "gender" = "male", "volume" = 50)
```

### 28. Find positions of TRUE elements using which()

```r
x = c(10, 15, 8, 14, 6, 12)
x

which(x == 14)
which(x != 12)
which(x > 10)
```

### 29. Find minimum and maximum positions in a matrix using which.min() and which.max()

```r
x = matrix(nrow = 3, ncol = 3, data = 1:9)
x

which.min(x)
which.max(x)
```

### 30. Find positions of odd values in a matrix using which()

```r
x = matrix(nrow = 3, ncol = 3, data = 1:9)
x

which(x %% 2 == 1)
which(x %% 2 == 1, arr.ind = TRUE)
```
