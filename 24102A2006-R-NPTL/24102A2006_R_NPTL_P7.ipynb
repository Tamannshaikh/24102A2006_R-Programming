## Lecture 29: Sorting, Ordering and Mode

### 27. Sort a vector in ascending and descending order

```r
y = c(8, 5, 7, 6)
y

sort(y)
sort(y, decreasing = TRUE)
```

### 28. Find the order of values in ascending and descending order

```r
y = c(9, 8, 5, 7, 6)
y

order(y)
order(y, decreasing = TRUE)
```

### 29. Find the mode of numeric and character objects

```r
mode(2.432)
mode(c(3, 4, 5, 6, 7, 8))
mode("India")
mode(c("India", "CANADA"))
```

### 30. Find the mode of a factor, list, data frame and function

```r
mode(factor(c("UP", "MP")))
mode(list("India", "USA"))
mode(data.frame(x = 1:2, y = c("India", "USA")))
mode(print)
```
