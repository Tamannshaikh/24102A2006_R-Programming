## Lecture 22: Loops – for loop

### 1. Use a for loop to print the squares of numbers from 1 to 5

```r
for (i in 1:5) {
  print(i^2)
}
```

### 2. Use a for loop with a vector of selected values

```r
for (i in c(2, 4, 6, 7)) {
  print(i^2)
}
```

### 3. Count values satisfying a condition using a for loop and a function

```r
x = c(2, 4, 6, 8, 10, 12)

excount = function(x) {
  count = 0
  for (xval in x) {
    if (xval / 2 > 3)
      count = count + 1
  }
  print(count)
}

excount(x)
```

### 4. Perform nested looping with two for loops

```r
child = c("child1", "child2", "child3")
sweet = c("sweet1", "sweet2", "sweet3")

for (x in child) {
  for (y in sweet) {
    print(paste(x, y))
  }
}
```

### 5. Use the break command to stop a for loop

```r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    break
  }
  print(x)
}
```

### 6. Use the next command to skip an iteration for lemonade

```r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "lemonade") {
    next
  }
  print(x)
}
```

### 7. Use the next command to skip an iteration for tea

```r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    next
  }
  print(x)
}
```

## Lecture 23: Loops – while and repeat

### 8. Use a while loop to print squares of odd numbers below 10

```r
i = 1

while (i < 10) {
  print(i^2)
  i = i + 2
}
```

### 9. Define a function using a while loop to calculate a sum

```r
sumfunction = function() {
  sum = 0
  number = as.integer(readline(prompt = "Please select any number less than 25: "))
  while (number <= 25) {
    sum = sum + number
    number = number + 1
  }
  print(paste("The sum of numbers received from the While Loop: ", sum))
}

# Example: enter 22 when prompted
sumfunction()
```

### 10. Verify the sum obtained from the while loop

```r
22 + 23 + 24 + 25
```

### 11. Use a repeat loop with the break command

```r
i = 1

repeat {
  print(i^2)
  i = i + 2
  if (i > 10)
    break
}
```

### 12. Use next and break together in a repeat loop

```r
i = 1

repeat {
  i = i + 1
  if (i < 10)
    next
  print(i^2)
  if (i >= 13)
    break
}
```

## Lecture 24: Functions

### 13. Define and call a function with one variable

```r
abc = function(x) {
  x^2
}

abc(3)
abc(6)
abc(9)
```

### 14. Define and call a function with two variables

```r
abc = function(x, y) {
  x^2 + y^2
}

abc(3, 4)
abc(10, 10)
abc(-2, -3)
```

### 15. Define a function using sin(), cos() and a variable

```r
abc = function(x) {
  sin(x)^2 + cos(x)^2 + x
}

abc(9)
abc(99)
abc(-15)
```

### 16. Define and call a function without an argument

```r
abc = function() {
  for (i in 1:3) {
    print(i^3)
  }
}

abc()
```
