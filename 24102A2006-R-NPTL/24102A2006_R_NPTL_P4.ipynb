## Difference Between `&` and `&&`

The NPTEL example shows that `&&` operates only on the first element, whereas `&` operates on all elements.


```r
# Define a vector
x = c(8,18)

# Display the vector
x

# Check each element using less than operator
x < 10

# Check each element using greater than operator
x > 2

# Scalar AND operator (&&)
# Use the first element explicitly because && works with a single logical value
(x[1] < 10) && (x[1] > 2)

# Element-wise AND operator (&)
# Checks the condition for each element of the vector
(x < 10) & (x > 2)
```

## Filtering Values Using `&`

Generate `x = 1:6`, check which values are greater than 2 and less than 5, and then extract those values.


```r
x = 1:6

(x > 2) & (x < 5)

x[(x > 2) & (x < 5)]
```

## Filtering Values Using `|`

Check which values are greater than 2 or less than 5 and extract the matching values.


```r
x = 1:6

(x > 2) | (x < 5)

x[(x > 2) | (x < 5)]
```

## Longer Logical Form on a Vector

The NPTEL example demonstrates that `&&` examines only the first element of each vector.


```r
# Create a vector from 1 to 6
x = 1:6

# Display the vector
x

# Check which elements are greater than 2
x > 2

# Check which elements are less than 5
x < 5

# Scalar AND operator (&&)
# Compare only the first element
(x[1] > 2) && (x[1] < 5)

# Element-wise AND operator (&)
# Compare each element of the vector
(x > 2) & (x < 5)
```
