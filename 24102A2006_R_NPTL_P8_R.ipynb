## Lecture 33 – Factors

### 17. Create a factor from categorical values

```r
gender = factor(c("Male", "Female", "Male", "Female"))
gender
levels(gender)
```

### 18. Factor with numeric values

```r
x = c(3, 4, 5, 6, 1, 2, 3, 3, 4, 4, 5, 6)
x
y = as.factor(x)
y
```

### 19. Factor with character values

```r
x = factor(c("lemonade", "lemonade", "juice", "lemonade", "water"))
x
```

### 20. Define a factor using levels and labels

```r
y = c(1, 4, 3, 5, 4, 2, 4)
possible.dieface = c(1, 2, 3, 4, 5, 6)
labels.dieface = c("one", "two", "three", "four", "five", "six")
facy = factor(y, levels = possible.dieface, labels = labels.dieface)
facy
```

## Lecture 34 – Factors: Class and Unclass

### 21. Check the class of different R objects

```r
class(9)
class("9")
class(print)
x = matrix(nrow=2, ncol=2, data=1:4)
class(x)
```

### 22. Use unclass() on a factor

```r
brands = c("A","A","B","B","B","B","C")
brands_fac = factor(brands)
brands_fac
unclass(brands_fac)
```

### 23. Convert factor codes to colors

```r
colours = c("blue", "green", "red")
colours[unclass(brands_fac)]
```

### 24. View underlying codes of a factor

```r
x = factor(c("lemonade", "lemonade", "juice", "lemonade", "water"))
unclass(x)
levels(x)
```

### 25. Change the order of factor levels

```r
x = factor(c("lemonade", "lemonade", "juice", "lemonade", "water"), levels=c("water", "juice", "lemonade"))
x
unclass(x)
levels(x)
```

### 26. Create an ordered factor

```r
income = ordered(c("high", "high", "low", "medium", "medium"), levels=c("low", "medium", "high"))
income
unclass(income)
```
